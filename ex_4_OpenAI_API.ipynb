**Traccia:** Trasforma un set di dati strutturati (come un dizionario Python) in una descrizione testuale completa.

**Obiettivo:** Convertire dati strutturati in una descrizione leggibile.

```python
import ollama
```

```python
input_text = "Scrivi la descrizione in modo semplice."

data = {
    "name": "Giovanni",
    "age": 30,
    "city": "Roma",
    "profession": "Ingegnere"
}

response = ollama.chat(
    model="llama3.2:3b",
    messages=[
        {
            "role": "system",
            "content": """
            Sei un assistente specializzato nella trasformazione
            di dati strutturati in descrizioni testuali.

            Utilizza tutte le informazioni presenti nel dizionario
            per creare una descrizione completa, chiara e naturale.

            Non inventare informazioni aggiuntive.
            Restituisci esclusivamente la descrizione.
            """
        },
        {
            "role": "user",
            "content": f"Trasforma questi dati in una descrizione testuale: {data}. Testo aggiuntivo: {input_text}"
        }
    ],
)

print(response.message.content)
```
