**Obiettivo:** Sviluppare un prompt che sia adatto alla traduzione fluida tra italiano e inglese.

```python

# TODO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env') # Read the .env file

# Configure OpenAI client
openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python

input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": """
            Sei un traduttore professionale.
            Traduci automaticamente dall'italiano all'inglese
            e dall'inglese all'italiano.

            Mantieni il significato originale, il tono e lo stile.
            Utilizza un linguaggio naturale, fluido e corretto.
            Non aggiungere spiegazioni o commenti.
            Restituisci esclusivamente il testo tradotto.
            """
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
)

print(complention.choices[0].message.content)
```
