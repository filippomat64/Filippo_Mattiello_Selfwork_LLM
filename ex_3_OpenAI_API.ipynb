# Esercizio 3

**Traccia:** Crea un prompt che riassuma un articolo di notizie in un singolo paragrafo di al massimo 255 caratteri.

**Obiettivo:** Prendere un testo lungo e condensarlo nei punti essenziali.

```python
import ollama
```

```python
input_text = "Apple ha presentato oggi un nuovo modello di iPhone. Il dispositivo introduce una fotocamera migliorata, una batteria più duratura e un processore più veloce. Il lancio sul mercato è previsto per settembre."

response = ollama.chat(
    model="llama3.2:3b",
    messages=[
        {
            "role": "system",
            "content": """
            Sei un assistente specializzato nel riassunto di notizie.

            Riassumi l'articolo in un unico paragrafo
            di massimo 255 caratteri, spazi inclusi.

            Mantieni solo le informazioni essenziali,
            utilizzando un linguaggio chiaro e preciso.

            Non aggiungere informazioni non presenti nel testo.
            Non superare mai i 255 caratteri.
            """
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
)

print(response.message.content)
```
