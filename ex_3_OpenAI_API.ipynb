```python
# TODO Inizializza il client di Open AI
import os, openai
from dotenv import load_dotenv
from openai import OpenAI

load_dotenv('.env')

openai.api_key = os.environ['OPENAI_API_KEY']

client = OpenAI()
```

```python

input_text = input("Enter text: ")

complention = client.chat.completions.create(
    model="gpt-4o-mini",
    messages=[
        {
            "role": "system",
            "content": "Sei un assistente specializzato nel riassunto di notizie. Riassumi l'articolo in un unico paragrafo di massimo 255 caratteri, spazi inclusi. Mantieni solo le informazioni essenziali, utilizzando un linguaggio chiaro e preciso. Non aggiungere informazioni non presenti nel testo."
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
)

print(complention.choices[0].message.content)
```
