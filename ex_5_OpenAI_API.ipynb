**Traccia:** Scrivi un prompt che generi domande di comprensione su un breve testo.

**Obiettivo:** Sviluppare un prompt per creare domande educative basate su un contenuto.

```python
import ollama
```

```python
input_text = "Leonardo da Vinci nacque nel 1452 ad Anchiano, vicino a Vinci, in Toscana. Fu un artista, inventore e scienziato italiano. Tra le sue opere più famose troviamo la Gioconda e l'Ultima Cena. Morì nel 1519 in Francia."

response = ollama.chat(
    model="llama3.2:3b",
    messages=[
        {
            "role": "system",
            "content": """
            Sei un insegnante specializzato nella creazione
            di domande di comprensione del testo.

            Leggi attentamente il testo fornito e genera
            esattamente 5 domande basate esclusivamente
            sulle informazioni presenti.

            Le domande devono essere chiare, pertinenti
            e di difficoltà variabile.

            Numera le domande da 1 a 5.
            Non fornire le risposte.
            """
        },
        {
            "role": "user",
            "content": input_text
        }
    ],
)

print(response.message.content)
```
