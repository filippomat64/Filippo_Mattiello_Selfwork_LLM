# Esercizio 2

**Traccia:** Crea un prompt che estragga i dettagli chiave (nome, età, professione) da un testo descrittivo su una persona.

**Obiettivo:** Estrarre informazioni strutturate da un testo libero.

```python

import ollama
```

```python
input_text = "Mi chiamo Marco, ho 28 anni e lavoro come ingegnere informatico."

response = ollama.chat(
    model="llama3.2:3b",
    messages=[
        {
            "role": "system",
            "content": """
            Sei un assistente specializzato nell'estrazione di informazioni.

            Analizza il testo fornito ed estrai esclusivamente:
            - Nome
            - Età
            - Professione

            Restituisci i risultati in formato strutturato,
            utilizzando un elenco puntato.

            Se un'informazione non è presente nel testo,
            indica "Non specificato".

            Non inventare informazioni e non aggiungere dettagli
            che non sono presenti nel testo.
            """
        },
        {
            "role": "user",
            "content": input_text
        }
    ]
)

print(response.message.content)
```
